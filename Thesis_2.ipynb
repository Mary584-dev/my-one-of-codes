```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
df = pd.read_csv('/content/OECD.STI.DEP,DSD_ICT_B@DF_BUSINESSES,1.0,complete,2026-10-11 19-47-26.csv')
df.head()
```

```python
df.isna().sum()
```

```python
df.drop(columns=['Time period','FREQ','REF_AREA','MEASURE','Observation value','ACTION','STRUCTURE_ID','OBS_STATUS_2','Observation status 2','OBS_STATUS_3','Observation status 3','TIME_HORIZON_USE','Time horizon'],axis= 1)
```

Several columns are dropped to remain with columns i need for analysis.

```python
dt = df[[
    "REF_AREA",
    "Economic activity",
    "MEASURE",
    "TIME_PERIOD",
    "OBS_VALUE"

]]

print(dt)
```

A portion from the main dataset was selected relevant for the study.

```python
df["MEASURE"].value_counts().plot.pie(autopct="%1.1f%%")

plt.show()
```

The pie chart shows that businesses mostlty use digital tools for online purchase.

```python
sns.barplot(data=df, x='Economic activity', y='OBS_VALUE')
plt.xlabel('Economic activity')
plt.ylabel('Observed value')
plt.title('Observed Values by Economic Activity')
plt.xticks(rotation=90)
plt.show()
```

Financial sectors is found to use digital technology

```python
activity_list = (
    df.groupby('Economic activity')['OBS_VALUE']
      .mean()
      .sort_values(ascending=False)
)

print(activity_list)
```

Different business activities are seen to be adopting technology differently
