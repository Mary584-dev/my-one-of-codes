# **Digitalisation in Businesses**

The dataset below is meant to analyse how different sectors, countries have adopted technology in different years.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
df = pd.read_csv('/content/OECD.STI.DEP,DSD_ICT_B@DF_BUSINESSES,1.0,complete,2026-10-11 19-47-26.csv')
df.head()
```

The first five rows of the dataset is displayed.

```python
df.isna().sum()
```

The dataset is found to have columns with null values

```python
df.drop(columns=['Time period','FREQ','REF_AREA','MEASURE','Observation value','ACTION','STRUCTURE_ID','OBS_STATUS_2','Observation status 2','OBS_STATUS_3','Observation status 3','TIME_HORIZON_USE','Time horizon'],axis= 1)
```

Several columns are dropped in order to remain with columns that are needed for analysis.

```python
dt = df[[
    "REF_AREA",
    "Economic activity",
    "MEASURE",
    "TIME_PERIOD",
    "OBS_VALUE"

]]

print(dt)
```

A portion from the main dataset was selected relevant for the study. OBS_VALUE is the percentage of ICT adoption in a company.Measure is different ways businesses have incoorporated ICT in their company.

```python
print(df['OBS_VALUE'].min())
print(df['OBS_VALUE'].max())
```

Minimum and maximum of OBS_VALUE is found to have values above 100%.Which is not logical and this to be dropped.

```python
count_above_100 = (df['OBS_VALUE'] > 100).sum()

print("Count of values above 100:", count_above_100)
```

The values above 100 are found to 1034.

```python
dt.dropna(inplace=True)
```

Null values are dropped

```python
dt = dt.drop(dt[dt['OBS_VALUE'] > 100].index)
```

Rows with values above 100% are dropped

```python
sns.barplot(data=dt, x='Economic activity', y='OBS_VALUE')
plt.xlabel('Economic activity')
plt.ylabel('Observed value')
plt.title('Observed Values by Economic Activity')
plt.xticks(rotation=90)
plt.show()
```

Financial sectors is found to use digital technology.

Different business activities are seen to be adopting technology differently

```python
activity_list = (
    dt.groupby('Economic activity')['OBS_VALUE']
      .mean()
      .sort_values(ascending=False)
)

print(activity_list)
```

```python
yearly_adoption = dt.groupby('TIME_PERIOD')['OBS_VALUE'].mean()

print(yearly_adoption)
```
